```python
import pandas as pd
```

```python
df = pd.read_csv("Telco_customer_churn.csv")
```

```python
# Converting Heading as snake_case

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df.columns
```

```python
df.info()
```

```python
# Check Missing Values

df.isnull().sum()
```

```python
# Percentage of Missing Values

(df.isnull().sum()/len(df)*100).sort_values(ascending= False)
```

```python
# Check Duplicate Rows

df.duplicated().sum()
```

```python
# Check Unique Values

for col in df.select_dtypes(include = 'object').columns:
    print(f"\n{col}")
    print(df[col].unique())
```

Note:

errors = 'coesrce'  --> Finds a number that cannot be converted into a number, it will replace it with NaN

```python
# Check Total Charges

df['total_charges'] = pd.to_numeric( df['total_charges'] , errors = 'coerce')
```

```python
df['total_charges']
```

```python
df['total_charges'].isnull().sum()
```

```python
# Handle Missing Values

df['total_charges'] = df['total_charges'].fillna(0)
```

```python
df['total_charges'].isnull().sum()
```
